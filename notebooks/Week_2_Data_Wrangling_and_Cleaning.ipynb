```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel("Online Retail.xlsx")

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
# Display basic information about the dataset

print("COLUMN NAMES:")
print(df.columns.tolist())

print("\nDATA TYPES:")
print(df.dtypes)

print("\nFIRST 5 ROWS:")
display(df.head())
```

```python
# Check missing values in each column

missing_values = df.isnull().sum()
missing_percentage = (missing_values / len(df) * 100).round(2)

missing_summary = pd.DataFrame({
    "Missing Values": missing_values,
    "Percentage (%)": missing_percentage
})

print("MISSING VALUES SUMMARY:")
display(missing_summary[missing_summary["Missing Values"] > 0])
```

```python
# Check for exact duplicate rows

duplicate_count = df.duplicated().sum()

print(f"Number of exact duplicate rows: {duplicate_count:,}")
print(f"Percentage of duplicate rows: {(duplicate_count / len(df) * 100):.2f}%")
```

```python
# Check transaction and value anomalies

cancelled_rows = df["InvoiceNo"].astype(str).str.startswith("C").sum()
negative_quantity = (df["Quantity"] < 0).sum()
zero_quantity = (df["Quantity"] == 0).sum()
negative_price = (df["UnitPrice"] < 0).sum()
zero_price = (df["UnitPrice"] == 0).sum()

print("DATA QUALITY CHECKS")
print("-" * 35)
print(f"Cancelled transaction rows: {cancelled_rows:,}")
print(f"Negative quantities: {negative_quantity:,}")
print(f"Zero quantities: {zero_quantity:,}")
print(f"Negative unit prices: {negative_price:,}")
print(f"Zero unit prices: {zero_price:,}")
```

```python
# Statistical summary for numerical columns

print("QUANTITY SUMMARY")
display(df["Quantity"].describe())

print("\nUNIT PRICE SUMMARY")
display(df["UnitPrice"].describe())
```

```python
# Display records with extreme Quantity or UnitPrice values

extreme_records = df[
    (df["Quantity"].abs() > 1000) |
    (df["UnitPrice"].abs() > 1000)
]

print(f"Number of potentially extreme records: {len(extreme_records)}")

display(
    extreme_records[
        ["InvoiceNo", "StockCode", "Description",
         "Quantity", "InvoiceDate", "UnitPrice",
         "CustomerID", "Country"]
    ].sort_values(["Quantity", "UnitPrice"])
)
```

```python
# Investigate records with zero or negative unit prices

unusual_prices = df[df["UnitPrice"] <= 0]

print(f"Number of records with zero or negative prices: {len(unusual_prices):,}")

display(
    unusual_prices[
        ["InvoiceNo", "StockCode", "Description",
         "Quantity", "InvoiceDate", "UnitPrice",
         "CustomerID", "Country"]
    ].head(20)
)
```

```python
# Create a cleaned dataset for sales analysis

df_clean = df.copy()

# 1. Remove exact duplicate rows
df_clean = df_clean.drop_duplicates()

# 2. Keep completed sales with positive quantities
df_clean = df_clean[df_clean["Quantity"] > 0]

# 3. Keep transactions with valid positive prices
df_clean = df_clean[df_clean["UnitPrice"] > 0]

# 4. Remove records without a product description
df_clean = df_clean.dropna(subset=["Description"])

print("CLEANING COMPLETED SUCCESSFULLY!")
print("-" * 40)
print(f"Original dataset rows: {len(df):,}")
print(f"Cleaned dataset rows: {len(df_clean):,}")
print(f"Rows removed: {len(df) - len(df_clean):,}")
print(f"Data retained: {(len(df_clean) / len(df) * 100):.2f}%")
```

```python
# Final validation of the cleaned dataset

print("FINAL DATA QUALITY VALIDATION")
print("-" * 40)

print(f"Duplicate rows: {df_clean.duplicated().sum()}")
print(f"Missing descriptions: {df_clean['Description'].isnull().sum()}")
print(f"Negative or zero quantities: {(df_clean['Quantity'] <= 0).sum()}")
print(f"Negative or zero prices: {(df_clean['UnitPrice'] <= 0).sum()}")

print("\nMissing values remaining:")
print(df_clean.isnull().sum())
```

```python
df_customers = df_clean.dropna(subset=["CustomerID"])
```

```python
# Create business analytics features

# Total revenue for each transaction
df_clean["TotalSales"] = df_clean["Quantity"] * df_clean["UnitPrice"]

# Extract time-based features from InvoiceDate
df_clean["Year"] = df_clean["InvoiceDate"].dt.year
df_clean["Month"] = df_clean["InvoiceDate"].dt.month
df_clean["MonthName"] = df_clean["InvoiceDate"].dt.month_name()
df_clean["Day"] = df_clean["InvoiceDate"].dt.day
df_clean["DayOfWeek"] = df_clean["InvoiceDate"].dt.day_name()

# Preview the transformed dataset
print("DATA TRANSFORMATION COMPLETED!")
print(f"Number of columns: {df_clean.shape[1]}")

display(df_clean.head())
```

```python
# Verify new columns and generate a revenue summary

print("NEW COLUMNS CREATED:")
print(df_clean.columns.tolist())

print("\nREVENUE SUMMARY")
print("-" * 40)
print(f"Total Revenue: £{df_clean['TotalSales'].sum():,.2f}")
print(f"Average Transaction Value: £{df_clean['TotalSales'].mean():,.2f}")
print(f"Maximum Transaction Value: £{df_clean['TotalSales'].max():,.2f}")
print(f"Minimum Transaction Value: £{df_clean['TotalSales'].min():,.2f}")
```

```python
# Save the cleaned and transformed dataset

df_clean.to_csv("online_retail_cleaned.csv", index=False)

print("Cleaned dataset saved successfully!")
print(f"Final dataset shape: {df_clean.shape}")
```
